Goal: expose our tools through the Model Context Protocol (MCP), the standard "USB port" for tools.
Any MCP client can then discover and call them, with no custom glue code.

## 1. Start the server and list its tools
`nka.mcp_server` is a separate process. The client below starts it, asks what it offers, and shuts it down.

```python
from nka.mcp_client import list_tools, call_tool, read_resource

for name, desc in list_tools():
    print(f"{name:12} {desc}")
print("documents behind the server:", read_resource("docs://count"))
```

## 2. Call tools through MCP

```python
print(call_tool("search_docs", query="hotel cap per night London", k=1)[:200])
print(call_tool("calculator", expression="12.5 / 100 * 8400"))
```
